```python
import numpy as np


def contar_elementos(x):
    # Se for um array NumPy, retorna o tamanho dele
    if isinstance(x, np.ndarray):
        return x.size

    # Se for nulo ou não for uma string ou lista, retorne 0
    # Note que pd.isna(x) não deve ser usada diretamente se x for um array!
    if pd.isna(x):
        return 0

    # Se for uma lista (real), conta seus elementos
    if isinstance(x, list):
        return len(x)

    # Se for uma string, processa-a
    if isinstance(x, str):
        texto = x.strip()  # remove espaços no início e fim
        if texto in ("[]", "[ ]"):  # caso de string vazia entre colchetes
            return 0
        # Remove os colchetes e espaços extras
        conteudo = texto.strip("[]").strip()
        if conteudo == "":
            return 0
        # Separa os elementos usando espaços (ajuste se os elementos forem separados por vírgulas)
        elementos = conteudo.split()
        return len(elementos)

    # Se não se enquadrar em nenhum dos casos anteriores, retorna 0
    return 0
```

```python
# Função segura que lida com np.array e valores nulos
def map_genres(id_array):
    if isinstance(id_array, (np.ndarray, list)):
        return [
            genres_group_map.get(i)
            for i in id_array
            if genres_group_map.get(i) is not None
        ]
    else:
        return None
```
