```python
# check installed version
import pycaret
pycaret.__version__
```

```python
# loading sample dataset from pycaret dataset module
from pycaret.datasets import get_data
data = get_data('diabetes')
```

```python
# import pycaret classification and init setup
from pycaret.classification import *

s = setup(
    data,
    target="Class variable",
    session_id=123,
    log_experiment="mlflow",
    experiment_name="diabetes_experiment",
    log_data=True,
    # log_profile=True,
    log_plots=True,
)
```

```python
# import ClassificationExperiment and init the class
from pycaret.classification import ClassificationExperiment
exp = ClassificationExperiment()
```

```python
# init setup on exp
exp.setup(data, target = 'Class variable', session_id = 123)
```

```python
# compare baseline models
best = compare_models(exclude=['lightgbm'])
```

```python
# compare models using OOP
exp.compare_models(exclude=['lightgbm'])
```

```python
# plot AUC
plot_model(best, plot = 'auc')
```

```python
evaluate_model(best)
```

```python
# predict on test set
holdout_pred = predict_model(best)
```

```python
# show predictions df
holdout_pred.head()
```

```python
# copy data and drop Class variable

new_data = data.copy()
new_data.drop('Class variable', axis=1, inplace=True)
new_data.head()
```

```python
# predict model on new_data
predictions = predict_model(best, data = new_data)
predictions.head()
```
